```python
import cv2
import numpy as np
from pathlib import Path
from PIL import Image, ImageDraw, ImageFont
import easyocr
import matplotlib.pyplot as plt


def put_chinese_text(img, text, position, font, color=(255, 0, 0), stroke_width=0):
    """Draw Chinese text on an OpenCV BGR image using Pillow."""
    img_pil = Image.fromarray(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    draw = ImageDraw.Draw(img_pil)
    fill_rgb = (color[2], color[1], color[0])
    draw.text(position, text, font=font, fill=fill_rgb, stroke_width=stroke_width, stroke_fill=fill_rgb)
    return cv2.cvtColor(np.array(img_pil), cv2.COLOR_RGB2BGR)


font_candidates = [
    Path(r"C:\Windows\Fonts\simhei.ttf"),
    Path(r"C:\Windows\Fonts\msyh.ttc"),
    Path(r"C:\Windows\Fonts\simsun.ttc"),
]
font_path = next((path for path in font_candidates if path.exists()), None)
if font_path is None:
    raise FileNotFoundError(
        "No Chinese font was found. Install SimHei, Microsoft YaHei, or SimSun."
    )
font = ImageFont.truetype(str(font_path), 24)

image_path = Path("test.jpg")
img = cv2.imread(str(image_path))
if img is None:
    raise FileNotFoundError(f"Could not load image: {image_path.resolve()}")

reader = easyocr.Reader(["ch_sim"], gpu=False)
results = reader.readtext(img)

for bbox, detected_text, score in results:
    print(bbox, detected_text, score)
    left, top = map(int, bbox[0])
    right, bottom = map(int, bbox[2])
    cv2.rectangle(img, (left, top), (right, bottom), (0, 255, 0), 2)
    img = put_chinese_text(
        img,
        detected_text,
        (left, max(0, top - 24)),
        font,
        color=(255, 0, 0),
        stroke_width=1,
    )

plt.figure(figsize=(12, 8))
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```
